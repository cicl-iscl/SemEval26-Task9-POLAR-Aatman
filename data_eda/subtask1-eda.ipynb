```python
from pathlib import Path

import pandas as pd
```

```python
train_dir = Path(
    "/home/aatman/Aatman/Study/Challenges in Computatinal Linguistics/code/data/subtask1/train"
)
dev_dir = Path(
    "/home/aatman/Aatman/Study/Challenges in Computatinal Linguistics/code/data/subtask1/dev"
)
test_dir = Path(
    "/home/aatman/Aatman/Study/Challenges in Computatinal Linguistics/code/data/subtask1/test"
)
```

```python
# Load all data
train_dfs = {f.stem: pd.read_csv(f) for f in train_dir.glob("*.csv")}
dev_dfs = {f.stem: pd.read_csv(f) for f in dev_dir.glob("*.csv")}
test_dfs = {f.stem: pd.read_csv(f) for f in test_dir.glob("*.csv")}
```

```python
# Combine for stats
train_all = pd.concat(train_dfs.values(), ignore_index=True)
dev_all = pd.concat(dev_dfs.values(), ignore_index=True)
test_all = pd.concat(test_dfs.values(), ignore_index=True)
```

## Overall Dataset Statistics

```python
print(f"Total train samples: {len(train_all):,}")
print(f"Total dev samples: {len(dev_all):,}")
print(f"Total test samples: {len(test_all):,}")
print(f"Number of languages: {len(train_dfs)}")
print("\nClass distribution in train:")
print(train_all["polarization"].value_counts().to_dict())
print(
    f"Class balance: {train_all['polarization'].value_counts(normalize=True).to_dict()}"
)
print("\nClass distribution in dev:")
print(dev_all["polarization"].value_counts().to_dict())
print(
    f"Class balance: {dev_all['polarization'].value_counts(normalize=True).to_dict()}"
)
print("\nTest dataset (no labels):")
print("Test samples have no polarization labels - used for final evaluation")
```

## Per-Language Statistics

```python
# Per-language stats
lang_stats = []
for lang in sorted(train_dfs.keys()):
    train_size = len(train_dfs[lang])
    dev_size = len(dev_dfs[lang])
    test_size = len(test_dfs[lang])

    # Train class distribution
    class_dist = train_dfs[lang]["polarization"].value_counts().to_dict()
    class_0 = class_dist.get(0, 0)
    class_1 = class_dist.get(1, 0)
    total = class_0 + class_1

    # Dev class distribution
    class_dist_dev = dev_dfs[lang]["polarization"].value_counts().to_dict()
    class_0_dev = class_dist_dev.get(0, 0)
    class_1_dev = class_dist_dev.get(1, 0)
    total_dev = class_0_dev + class_1_dev

    lang_stats.append(
        {
            "language": lang,
            # Train stats
            "train": train_size,
            "train_class_0": class_0,
            "train_class_1": class_1,
            "train_pct_class_0": class_0 / total if total > 0 else 0,
            "train_pct_class_1": class_1 / total if total > 0 else 0,
            # Dev stats
            "dev": dev_size,
            "dev_class_0": class_0_dev,
            "dev_class_1": class_1_dev,
            "dev_pct_class_0": class_0_dev / total_dev if total_dev > 0 else 0,
            "dev_pct_class_1": class_1_dev / total_dev if total_dev > 0 else 0,
            # Test stats (no labels)
            "test": test_size,
        }
    )

stats_df = pd.DataFrame(lang_stats)
stats_df
```

## Text Length Analysis

```python
# Add text length columns
train_all["text_length"] = train_all["text"].str.len()
dev_all["text_length"] = dev_all["text"].str.len()
test_all["text_length"] = test_all["text"].str.len()

print("Text length statistics:")
print("\nTrain set:")
print(train_all["text_length"].describe())
print("\nDev set:")
print(dev_all["text_length"].describe())
print("\nTest set:")
print(test_all["text_length"].describe())
```
