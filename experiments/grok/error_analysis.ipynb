```python
import pandas as pd
from sklearn.metrics import f1_score, accuracy_score
```

```python
gold_df = pd.read_csv(r'/home/aatman/Aatman/Study/Challenges in Computatinal Linguistics/code/data/subtask1/dev/hin.csv', lineterminator='\n')
gold_df.shape
```

```python
gold_df.head()
```

```python
grok_df = pd.read_csv(r'hin_grok_output.csv')
grok_df.shape
```

```python
grok_df.head()
```

```python
# check if text is same in both
(gold_df['text'] == grok_df['text']).all()
```

```python
merged = gold_df.merge(
    grok_df,
    on='text',
    how='inner'
)
merged.shape
```

```python
merged['grok_label'] = merged['polarized'].map({
    'Yes': 1,
    'No': 0
})
```

```python
merged
```

```python
merged[['polarization\r', 'polarized', 'grok_label']].head()
```

```python
y_true = merged['polarization\r']
y_pred = merged['grok_label']

f1_macro = f1_score(y_true, y_pred, average='macro')
accuracy = accuracy_score(y_true, y_pred)

f1_macro, accuracy
```
